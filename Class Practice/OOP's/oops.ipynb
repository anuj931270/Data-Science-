# The __init__() Method
All classes have a built-in method called __init__(), which is always executed when the class is being initiated.

The __init__() method is used to assign values to object properties, or to perform operations that are necessary when the object is being created.

__init__() is a constructor. It runs automatically when an object is created

```python
class MyClass: #Each object is independent and has its own copy of the class properties.
  x = 5

p1 = MyClass()
p2 = MyClass()
p3 = MyClass()


print(p1.x)
print(p2.x)
print(p3.x)
```

```python
# Create a class called Dog
# Add an __init__ method with parameters name and age, and store them as properties using self
# Add a method called bark that prints the dog's name followed by " says Woof!"
# Create an object d1 of the Dog class with name "Buddy" and age 3
# Call the bark method on d1
 
class Dog:                                  #Create a Class and the name is Dog
    def __init__(self,name,age):            #Constructor (__init__)
        self.name=name                      #Store Values in Object Properties
        self.age=age
    def bark(self):
        print(self.name, "says Woof!")
d1=Dog("Buddy", 3)                          #we create an object
d1.bark()
```

```python
# Create a class called Person
# Add an __init__ method that takes name and age as parameters
# Add a method called greet that prints "Hello, my name is " followed by the name
# Create an object p1 of the class with name "John" and age 36
# Call the greet method on p1


class Person:                      #Create a Class
    def __init__(self, name, age): #Constructor (__init__)
        self.name = name           #self.name stores the name.
        self.age = age             #self.age stores the age.

    def greet(self):               #This method prints the person's name.
        print("Hello, my name is", self.name) 


p1 = Person("John", 36) #Create an Object

p1.greet() #Call the Method
```

```python
class Bank_v1:
    bank_name = "ABC Bank"

    def __init__(self, name, balance, cibil_score):
        self.name = name
        self.balance = balance
        self.cibil_score = cibil_score

    def account_details(self):
        return f"Customer Name : {self.name}\nBalance : ₹{self.balance}"


class Bank_v2(Bank_v1):

    def withdraw(self, amount):
        if self.balance >= 5000:
            return "Minimum Balance Maintained. No Extra Charges."
        elif self.balance >= amount:
            self.balance -= amount
            return f"₹{amount} Withdrawn Successfully.\nRemaining Balance : ₹{self.balance}"
        else:
            return "Insufficient Balance."

    def rupay_card(self):
        return "RuPay Debit Card Activated Successfully."

    def debit_card(self):
        return "Debit Card Available."

    def upi(self):
        return "UPI Service Activated."

    def net_banking(self):
        return "Net Banking Activated Successfully."


class Loan_v1(Bank_v2):

    def loan_eligibility(self):
        if self.cibil_score >= 750:
            return "Eligible for Home Loan, Car Loan and Personal Loan."
        else:
            return "Not Eligible for Loan."

    def apply_loan(self, amount):
        if self.cibil_score >= 750:
            return f"Loan of ₹{amount} Approved."
        else:
            return "Loan Application Rejected."


class Card_v1(Loan_v1):

    def credit_card(self):
        if self.cibil_score >= 700:
            return "Credit Card Approved."
        else:
            return "Credit Card Rejected."

    def premium_card(self):
        if self.balance >= 100000:
            return "Eligible for Premium RuPay Credit Card."
        else:
            return "Not Eligible for Premium Card."


class Bank_System(Card_v1):
    pass


# Object Creation
customer = Bank_System("Anuj", 120000, 780)

print(customer.account_details())
print(customer.withdraw(2000))
print(customer.loan_eligibility())
print(customer.apply_loan(500000))

print(customer.rupay_card())
print(customer.debit_card())
print(customer.credit_card())
print(customer.premium_card())

print(customer.upi())
print(customer.net_banking())
```

```python
# Student Manage Sysytem
# main class student
# add stu features
# get all stu features
# get stu by roll no features
# update stu features
# delete student features


class Student:
    def __init__(self):
        self.students = {}

    def add_student(self, roll_no, name, age, city):
        if roll_no in self.students:
            print(f"Student with Roll No {roll_no} already exists.")
            return

        self.students[roll_no] = {
            "name": name,
            "age": age,
            "city": city
        }
        print(f"Student {name} added successfully.")

    def get_all_students(self):
        if not self.students:
            print("No Students Found.")
            return

        for roll_no, details in self.students.items():
            print(f"Roll No : {roll_no} | Name : {details['name']} | Age : {details['age']} | City : {details['city']}")

    def get_student_by_roll(self, roll_no):
        if roll_no not in self.students:
            print("Student Not Found.")
            return

        details = self.students[roll_no]
        print(f"Roll No : {roll_no}")
        print(f"Name    : {details['name']}")
        print(f"Age     : {details['age']}")
        print(f"City    : {details['city']}")

    def update_student(self, roll_no, name, age, city):
        if roll_no not in self.students:
            print("Student Not Found.")
            return

        self.students[roll_no] = {
            "name": name,
            "age": age,
            "city": city
        }

        print("Student Updated Successfully.")

    def delete_student(self, roll_no):
        if roll_no not in self.students:
            print("Student Not Found.")
            return

        del self.students[roll_no]
        print("Student Deleted Successfully.")
```

```python
# Object Creation
s = Student()
```

```python
# Add Students
s.add_student(100, "Aman", 33, "Noida")
s.add_student(101, "Rahul", 22, "Delhi")
s.add_student(102, "Priya", 21, "Lucknow")
```

```python
print("\nAll Students") #new line
s.get_all_students()
```

```python
print("\nStudnet By Roll No")
s.get_student_by_roll(102)
```

```python
print("\nUpdate Student")
s.update_student(102,"Priya Pathak",22,"Gurgaon")
s.get_student_by_roll(102)
```

```python
print("\nDelete Student")
s.delete_student
```
