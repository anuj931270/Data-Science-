```python
#Polymorphism
#1. Methid overriding

class Payment:
    def pay(self,amount):
        print("Genric Payment Method")

class Upi_payment(Payment):
    def pay(self,amount):
        print(F"Paid {amount} using UPI METHOD")

class Card_payment(Payment):
    def pay(self,amount):
        print(F"Paid {amount} using CARD METHOD")
class Net_payment(Payment):
    def pay(self,amount):
        print(F"Paid {amount} using NET METHOD")

for method in [Upi_payment(),Card_payment(),Net_payment()]:
    method.pay(1000)
        
```

```python
#2. Duck Typing : if walks like a duck and quacks like a duck its is a duck.

class Dog:
    def sound(self):
        print("Bark")
class Cat:
    def sound(self):
        print("Meow")

for animal in [Cat(),Dog()]:
    animal.sound()
```

```python
class Calculator:
    def add(self,a,b):
        return a+b
    def add(self,a,b,c):
        return a+b+c

c=Calculator()
c.add(2,3,5)
```

```python
#Default para
class Calculator:
    def add(self,*args):
        return sum(args)

c=Calculator()
print(c.add(2,3))
print(c.add(2,3,5))
print(c.add(2,3,5,6))
```

```python
#Type Checking
class Printer:
    def show(self,value):
        if isinstance(value,str):
            print("String : {value}")
        elif isinstance(value,int):
            print(f"Number : {value}")
        else:
            print("Method not definend...")

p=Printer()
p.show([4,4])
```

```python
class Person:
    def __init__(self, name, city, age):
        self.name = name
        self.city = city
        self.age = age


class Student(Person):
    def __init__(self, name, roll_no, city, age):
        super().__init__(name, city, age)
        self.roll_no = roll_no


s = Student("Aman", 111, "Noida", 25)

print(s.name)
print(s.roll_no)
print(s.city)
print(s.age)
```

```python
class A:
    def show(self):
        print("A")


class B(A):
    def show(self):
        super().show()
        print("B")
class C(B):
    def show(self):
        super().show()
        print("C")

class D(B,C):
    def show(self):
        super().show()
        print("C")

```

```python
#Classmethod

class Student:
    school_name="DevOps Academy"

    def __int__(self,name):
        self.name=name
    @classmethod
    def change_school(cls,new_name):
        cls.school_name=new_name

Student.change_school("Data Science Academy")
print(Student.school_name)
```

```python
#File management system
#main class File_system
# Dedicated directory FMS
import os

#1. Create file

class File_System:
    def __init__(self):
        self.folder="FMS"

        #create a floder
        if not os.path.exists(self.folder):
            os.mkdir(self.folder)
            print("FMS folder created successfully.")
        else:
            print("FMS folder already exists.")

    #Create File

    def create_file(self):
        file_name=("Enter file name")
        file_path=os.path.join(self.folder, file_name + ".txt")

        if os.path.exists(file_path):
            print("File already exists.")
        else:
            with open (file_path, "w") as file:
                data= input(("Enter file content: "))
                file.write(data)
            print("File created successfully.")


obj=File_System()
while True:
    print("\n===== FILE MANAGEMENT SYSTEM =====")
    print("1. Create File")
    print("2. Exit")

    choice = input("Enter Your Choice: ")

    if choice == "1":
        obj.create_file()

    elif choice == "2":
        print("Thank You!")
        break

    else:
        print("Invalid Choice!")


#2. Update Life
#3. Read all files
#4. Delete files
```
