It is built on top NumPy.

Scipy provides ready made mathmatical functions.

```python
#Module 2 - scipy.cluster
#used fot clustering data
import numpy as np
from sklearn.cluster import KMeans

#create data set
data = np.array([
    [1,2],
    [2,3],
    [3,4],
    [8,9],
    [9,10],
    [10,11],
    [11,12]
])

#We create a K-Means model
model = KMeans(n_clusters=2, random_state=42)

#train the model
model.fit(data)

print("Centroids:")
print(model.cluster_centers_)

print("\nLabels:")
print(model.labels_)
```

```python
# Module 3- scipy.stats
# Mean
# Median
# Mode
# Standard Deviation
# Correlation
# Probability
#SciPy helps me calculate statistical values quickly.
```

```python
#Mean 
from scipy import stats
data =[10,20,40,30]
print(stats.tmean(data))
```

```python
#Mode 
data= [10,30,50,70,10]
print(stats.mode(data))
```

```python
#Median

data= [10,50,20,30,5]
stats.median_abs_deviation(data)
```

```python
#Standard deviation
marks = [10,20,30,40,50]

print(stats.tstd(marks))
```

```python
#Variance

data= [10,25,35,45]
stats.tvar(data)
```

```python
# Find the mean of five numbers.

marks=[5,8,7,9,6,4]
print(stats.tmean(marks))
```

```python
# Module 4- scipy.linalg
#Matrix
from scipy import linalg
A= np.array([
    [1,2,5],
    [2,3,6],
    [3,4,8]
])
A
print(linalg.det(A))
```

```python
#Inverse Matrix
A= np.array([
    [1,2,5],
    [2,3,6],
    [3,4,8]
])

linalg.inv(A)
```

```python
#Eigen Values
#Linear algebra is important in machine learning. SciPy makes matrix caluclations easy.

A= np.array([
    [1,2,5],
    [2,3,6],
    [3,4,8]
])

linalg.eigvals(A)
```

```python
# Module 5- scipy.optimize

from scipy.optimize import minimize
# Suppose (x**2+5)
# we want the minimum value

def fun(x):
    return x**2+5

result=minimize(fun, x0=10)
result.x
```

```python
#example of root 
from scipy.optimize import root
def equation(x):
    return x**2-25
r=root(equation,0)
r.x
```

```python
#solve x²-16=0
def solution(x):
    return x**2-16
eq=root(solution,0)
eq.x
```

```python
# Module 6 — scipy.integrate
# used for integration
from scipy.integrate import quad

result,error= quad(lambda x:x**2,0,5)
result
```

```python
#Module 7 — scipy.spatial
#Distance caluclation used in Recommendation Systems
# KNN
# Clustering

from scipy.spatial.distance import euclidean

a=[4,5]
b=[5,6]
euclidean(a,b)
```

```python
#Used for predicting missing values in the given datasets.
#Module 8 — scipy.interpolate

from scipy.interpolate import interp1d
import numpy as np
x=np.array([1,2,3])
y=np.array([10,20,30])
f=interp1d(x,y)
f(2.1)
```

```python
# Module 9 — scipy.fft
# Fast Fourier Transform
# Mostly used in
# Audio Processing
# Image Processing
# Signal Processing

from scipy.fft import fft
import numpy as np
x=np.array([1,2,3,4])
print(fft(x))
```
