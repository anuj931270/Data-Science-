```python
# 19. Write a function to find the most repeated character in a string.
def repeated_char(text):
    max_char = ""
    max_count = 0

    for i in text:
        if i != " ":
            count = text.count(i)
            if count > max_count:
                max_count = count
                max_char = i

    print("Most repeated character:", max_char)
    print("Count:", max_count)

text = "python programming"
repeated_char(text)
```

```python
# 20. Write a function to convert all lowercase letters to uppercase without using .upper().
def to_uppercase(text):
    result = ""

    for i in text:
        if 'a' <= i <= 'z':
            result += chr(ord(i) - 32)
        else:
            result += i

    return result

text = "Python is Easy"
print(to_uppercase(text))
```

```python
# 27. Write a function to sort a list without using the built-in sort() function.
def sort_list(numbers):
    n = len(numbers)

    for i in range(n):
        for j in range(0, n - i - 1):
            if numbers[j] > numbers[j + 1]:
                # Swap the elements
                numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

    return numbers

numbers = [45, 12, 78, 34, 23, 89, 10]
print("Sorted List:", sort_list(numbers))
```

```python
# 28. Write a function to find the smallest string from a list of strings.
def smallest_string(words):
    smallest = words[0]

    for word in words:
        if len(word) < len(smallest):
            smallest = word

    return smallest

words = ["Apple", "Banana", "Kiwi", "Orange", "Mango"]
print("Smallest string:", smallest_string(words))
```

```python
# 34. Write a function to check if a number is an Armstrong number.
def armstrong_num(num):
    original = num
    digits = len(str(num))
    total = 0

    while num > 0:
        digit = num % 10
        total += digit ** digits
        num = num // 10

    if total == original:
        print("Armstrong Number")
    else:
        print("Not an Armstrong Number")

num = int(input("Enter a number: "))
armstrong_num(num)
```

```python
# 46. Write a function that generates a random password of 8 characters using letters and digits.
import random
import string
def generate_password():
    characters = string.ascii_letters + string.digits
    password = ""

    for i in range(8):
        password += random.choice(characters)

    return password

print("Random Password:", generate_password())
```

```python
# 47. Write a function to count how many numbers in a list are greater than the average of the list.
def greater_than_avg(numbers):
    total = 0

    for num in numbers:
        total += num

    average = total / len(numbers)

    count = 0
    for num in numbers:
        if num > average:
            count += 1

    return average, count

numbers = [10, 20, 30, 40, 50]

average, count = greater_than_avg(numbers)

print("Average =", average)
print("Numbers greater than average =", count)
```

```python
# 48. Write a function to display a menu-based simple calculator using while True loop and
# functions for each operation.
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Division by zero is not possible."
    return a / b


while True:
    print("\n===== Simple Calculator =====")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    if choice == "5":
        print("Calculator Closed.")
        break

    if choice in ["1", "2", "3", "4"]:
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))

        if choice == "1":
            print("Result =", add(num1, num2))

        elif choice == "2":
            print("Result =", subtract(num1, num2))

        elif choice == "3":
            print("Result =", multiply(num1, num2))

        elif choice == "4":
            print("Result =", divide(num1, num2))

    else:
        print("Invalid choice! Please enter a number between 1 and 5.")
```

```python
# 49. Write a function to return greatest substring in any sentence pass by user.
def greatest_substring(sentence):
    words = sentence.split()
    longest = words[0]

    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest


sentence = input("Enter a sentence: ")
print("Greatest substring:", greatest_substring(sentence))
```
