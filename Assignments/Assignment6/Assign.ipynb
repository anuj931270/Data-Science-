```python
# Write a program to calculate the sum of all keys in the dictionary
# d = {1: 1, 2: 2, 3: 3, 4: 4}

d = {1: 1, 2: 2, 3: 3, 4: 4}
total_sum=0

for i in d:
    total_sum+=i
print("The sum of all keys is :",total_sum)
```

```python
# Write a program to calculate the sum of both keys and values in the dictionary
# d = {1: 1, 2: 2, 3: 3, 4: 4}

# Define the dictionary
d = {1: 1, 2: 2, 3: 3, 4: 4}

# Calculate the sum of keys and the sum of values, then add them together
total_sum = sum(d.keys()) + sum(d.values())

# Print the result
print(f"The sum of both keys and values is: {total_sum}")
```

```python
# Create an empty dictionary called user_data.
# Allow the user to enter key-value pairs until they choose to stop.
# Print the final dictionary.

# Create an empty dictionary
user_data = {}

while True:
    # Take key and value from the user
    key = input("Enter key: ")
    value = input("Enter value: ")

    # Store in dictionary
    user_data[key] = value

    # Ask if the user wants to continue
    choice = input("Do you want to add another key-value pair? (yes/no): ").lower()

    if choice == "no":
        break

# Print the final dictionary
print("\nFinal Dictionary:")
print(user_data)
```

```python
# Write a program to calculate the total score of all students
# student_score = {1: 44, 2: 45, 3: 55}

student_score = {1: 44, 2: 45, 3: 55}

total_sum = 0

for score in student_score.values():
    total_sum += score

print("The total score of all students is:", total_sum)
```

```python
# Write a program to merge two dictionaries
# d1 = {1: "a", 2: "b"}
# d2 = {3: "c", 4: "d"}

d1 = {1: "a", 2: "b"}
d2 = {3: "c", 4: "d"}

merged={**d1,**d2}
print(merged)
```

```python
# Write a program to swap keys and values in the dictionary
# d = {1: "one", 2: "two", 3: "three"}

d = {1: "one", 2: "two", 3: "three"}
swapped={}
for key,value in d.items():
    swapped[value]=key
print("Swapped dictionary",swapped)
```

```python
# Write a program to count the frequency of each character in a string using a
# dictionary.
# Example: "banana"

string = "banana"
freq = {}

for char in string:
    freq[char] = freq.get(char, 0) + 1

print("Character frequencies:", freq)
```

```python
# Write a program to create a dictionary where keys are numbers from 1 to 5 and
# values are their squares.

squares = {}

for i in range(1, 6):
    squares[i] = i * i

print("Squares dictionary:", squares)
```

```python
# Write a program to sort a dictionary by its keys
# d = {3: "three", 1: "one", 2: "two"}
d = {3: "three", 1: "one", 2: "two"}

sorted_dict = {}
for key in sorted(d.keys()):
    sorted_dict[key] = d[key]

print("Sorted by keys:", sorted_dict)
```

```python
# Write a program to count how many values are greater than 50 in a dictionary.

data = {"a": 20, "b": 60, "c": 45, "d": 80, "e": 55}

count = 0
for value in data.values():
    if value > 50:
        count += 1

print("Values greater than 50:", count)
```
