```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

actual = [1,1,1,1,0,0,0,0,1,0]

predicted = [1,1,0,1,1,0,0,0,1,0]

cm = confusion_matrix(actual, predicted)

print(cm)
disp = ConfusionMatrixDisplay(cm)

disp.plot()

plt.show()
```

```python
#Accuracy
## Accuracy works well when dataset is balanced
## Best for: Balanced datasets Not suitable for: Imbalanced datasets (e.g., fraud detection, disease detection, spam filtering)

from sklearn.metrics import accuracy_score

actual = [1,1,1,1,0,0,0,0,1,0]

predicted = [1,1,0,1,1,0,0,0,1,0]

accuracy = accuracy_score(actual, predicted)

print(accuracy)
print(accuracy*100)
```

```python
#Recall Score
from sklearn.metrics import recall_score

actual = [1,1,1,1,0,0,0,0,1,0]

predicted = [1,1,0,1,1,0,0,0,1,0]

recall = recall_score(actual, predicted)

print(recall)
print(accuracy*100)
```

```python
#Precison
from sklearn.metrics import precision_score

actual = [1,1,1,1,0,0,0,0,1,0]

predicted = [1,1,0,1,1,0,0,0,1,0]

precision = precision_score(actual, predicted)

print(precision)
```
